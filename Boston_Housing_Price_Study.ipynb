```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
file = "housing.csv"

if not os.path.exists(file):
    raise FileNotFoundError("Error: housing.csv file not found!")

df = pd.read_csv(file)  
df.columns = df.columns.str.strip()

print("Housing Dataset")
print(df.head()) 
print("\nDataset Information:")
print(df.info()) 
print("\nMissing Values:")
print(df.isnull().sum()) 
df = df.dropna() 
required_columns = [
    "Avg. Area Income",
    "Avg. Area House Age",
    "Avg. Area Number of Rooms",
    "Avg. Area Number of Bedrooms",
    "Area Population",
    "Price"
]

if not all(col in df.columns for col in required_columns):
    raise ValueError("Required columns are missing!")

df = df[required_columns] 
print("\nStatistical Summary:")
print(df.describe()) 
plt.figure(figsize=(8, 5))
plt.hist(df["Price"], bins=20, color="skyblue",
         edgecolor="black")
plt.xlabel("House Prices")
plt.ylabel("Frequency")
plt.title("Distribution of House Prices")
plt.show() 
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Avg. Area Number of Rooms"],
            df["Price"], color="green")
plt.xlabel("Average Number of Rooms")
plt.ylabel("House Prices")
plt.title("Rooms vs House Prices")
plt.show() 
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Avg. Area Income"],
            df["Price"], color="blue")
plt.xlabel("Average Area Income")
plt.ylabel("House Prices")
plt.title("Income vs House Prices")
plt.show() 
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Avg. Area House Age"],
            df["Price"], color="orange")
plt.xlabel("Average House Age")
plt.ylabel("House Prices")
plt.title("House Age vs House Prices")
plt.show() 
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True,
            cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show() 
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Price"], color="lightgreen")
plt.xlabel("House Prices")
plt.title("Boxplot of House Prices")
plt.show() 
sns.pairplot(df.sample(min(500, len(df)), random_state=42))
plt.suptitle("Pairwise Relationships in Housing Dataset",
             y=1.02)
plt.show() 
print("\nExploratory Data Analysis Completed Successfully!") 
```
